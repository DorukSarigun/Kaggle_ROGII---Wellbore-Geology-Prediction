```python
from training import (BASELINE_FEATURE_COLUMNS, build_training_dataset, make_baseline_feature_functions)
from utils import get_all_test_ids, get_all_training_ids, train_folder

all_training_ids = get_all_training_ids()
test_ids = get_all_test_ids()

train_well_ids = [well_id for well_id in all_training_ids if well_id not in test_ids]
valid_well_ids = test_ids

assert not set(train_well_ids) & set(valid_well_ids)
```

```python
X_train, y_train, train_groups = build_training_dataset(
    well_ids=train_well_ids,
    feature_columns=BASELINE_FEATURE_COLUMNS,
    feature_function_factory=make_baseline_feature_functions,
    data_dir=train_folder,
    post_ps_only=False,
)

# Kaggle metriği yalnız gerçek PS sonrası satırları skorlar.
X_valid, y_valid, valid_groups = build_training_dataset(
    well_ids=valid_well_ids,
    feature_columns=BASELINE_FEATURE_COLUMNS,
    feature_function_factory=make_baseline_feature_functions,
    data_dir=train_folder,
    post_ps_only=True,
)
```

```python
from lightgbm import LGBMRegressor

model = LGBMRegressor(
    objective="regression",
    n_estimators=600,
    learning_rate=0.035,
    num_leaves=40,
    random_state=42,
    n_jobs=-1,
    verbosity=-1
)
```

```python
model.fit(X_train, y_train)
```

```python
pred_valid_tvt = model.predict(X_valid)
```

```python
from sklearn.metrics import root_mean_squared_error

valid_rmse = root_mean_squared_error(
    y_valid,
    pred_valid_tvt,
)

print(f"Validation RMSE: {valid_rmse:.4f}")
```

```python
import numpy as np

mean_tvt_baseline_rmse = root_mean_squared_error(
    y_valid,
    np.full(len(y_valid), y_train.mean()),
)

print(f"Mean TVT baseline RMSE: {mean_tvt_baseline_rmse:.4f}")
print(f"LightGBM RMSE:          {valid_rmse:.4f}")
```

```python
feature_importance = pd.Series(model.booster_.feature_importance(importance_type="gain"),
                                index=X_train.columns).sort_values(ascending=False)
```

```python
feature_importance
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor


hist_model = HistGradientBoostingRegressor(
    loss="squared_error",
    learning_rate=0.035,
    max_iter=600,
    max_leaf_nodes=40,
    l2_regularization=1.0,
    early_stopping=False,
    random_state=42,
)
```

```python
hist_model.fit(X_train, y_train)

hist_pred_valid_tvt = hist_model.predict(X_valid)

hist_rmse = root_mean_squared_error(
    y_valid,
    hist_pred_valid_tvt,
)

print(f"HistGradientBoosting RMSE: {hist_rmse:.4f}")
print(f"LightGBM RMSE:              {valid_rmse:.4f}")
```

## Catboost


```python
from catboost import CatBoostRegressor

cat_model = CatBoostRegressor(
    loss_function="RMSE",
    eval_metric="RMSE",
    iterations=1500,
    learning_rate=0.04,
    depth=8,
    l2_leaf_reg=3.0,
    random_seed=42,
    thread_count=-1,
    verbose=100,
    allow_writing_files=False
)
```

```python
cat_model.fit(X_train, y_train)
```

```python
cat_pred_valid_tvt = cat_model.predict(X_valid)

cat_rmse = root_mean_squared_error(
    y_valid,
    cat_pred_valid_tvt,
)

print(f"Global LightGBM RMSE: {valid_rmse:.4f}")
print(f"CatBoost RMSE: {cat_rmse:.4f}")
```

```python
cat_feature_importance = pd.Series(
    cat_model.get_feature_importance(),
    index=X_train.columns
).sort_values(ascending=False)
```
